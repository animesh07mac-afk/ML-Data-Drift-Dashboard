# 02 — Preprocessing & Feature Engineering
Clean datasets, detect feature types, and extract datetime features.

```python
import sys
sys.path.insert(0, '../src')

import pandas as pd
from preprocessing import (
    preprocess, get_feature_types,
    detect_numerical_features, detect_categorical_features, detect_datetime_features
)
from feature_engineering import run_feature_engineering
```

## 1. Load Raw Data

```python
ref_raw = pd.read_csv('../data/reference.csv')
cur_raw = pd.read_csv('../data/current.csv')
print('Shapes — Ref:', ref_raw.shape, '| Cur:', cur_raw.shape)
```

## 2. Preprocessing

```python
ref_df = preprocess(ref_raw)
cur_df = preprocess(cur_raw)
print('After preprocessing — Ref:', ref_df.shape, '| Cur:', cur_df.shape)
```

## 3. Feature Type Detection

```python
feature_types = get_feature_types(ref_df)
print('Numerical  :', feature_types['numerical'])
print('Categorical:', feature_types['categorical'])
print('Datetime   :', feature_types['datetime'])
```

## 4. Feature Engineering (Datetime Extraction)

```python
ref_eng = run_feature_engineering(ref_df, feature_types['datetime'])
cur_eng = run_feature_engineering(cur_df, feature_types['datetime'])
print('New columns added:', set(ref_eng.columns) - set(ref_df.columns))
ref_eng.head()
```

## 5. Final Feature Types After Engineering

```python
ft_final = get_feature_types(ref_eng)
print('Numerical  :', ft_final['numerical'])
print('Categorical:', ft_final['categorical'])
```
