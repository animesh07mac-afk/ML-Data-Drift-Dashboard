# 01 — Data Validation
Validate Reference and Current datasets: shape, schema, missing values, and duplicates.

```python
import sys
sys.path.insert(0, '../src')

import pandas as pd
from validation import (
    load_dataset, get_shape_info, compare_columns,
    compare_dtypes, missing_values_summary,
    duplicate_rows_summary, validate_schema, run_validation
)
```

## 1. Load Datasets

```python
ref_df = load_dataset('../data/reference.csv')
cur_df = load_dataset('../data/current.csv')
print('Reference:', ref_df.shape)
print('Current  :', cur_df.shape)
```

## 2. Dataset Shapes

```python
print(get_shape_info(ref_df, 'Reference'))
print(get_shape_info(cur_df, 'Current'))
```

## 3. Column Comparison

```python
compare_columns(ref_df, cur_df)
```

## 4. Data Types Comparison

```python
compare_dtypes(ref_df, cur_df)
```

## 5. Missing Values

```python
ref_miss = missing_values_summary(ref_df, 'Reference')
cur_miss = missing_values_summary(cur_df, 'Current')
print('Reference missing values:')
print(ref_miss if not ref_miss.empty else 'None')
print('\nCurrent missing values:')
print(cur_miss if not cur_miss.empty else 'None')
```

## 6. Duplicate Rows

```python
print(duplicate_rows_summary(ref_df, 'Reference'))
print(duplicate_rows_summary(cur_df, 'Current'))
```

## 7. Full Schema Validation

```python
schema = validate_schema(ref_df, cur_df)
print('Schema valid:', schema['schema_valid'])
schema['dtype_comparison']
```

## 8. Full Validation Report

```python
report = run_validation(ref_df, cur_df)
for key, val in report.items():
    print(f'--- {key} ---')
    print(val)
    print()
```
